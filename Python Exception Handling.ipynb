```python
balance = 5000

try:
    amount = float(input("Enter withdrawal amount: "))

    if amount <= 0:
        raise ValueError("Amount must be greater than zero.")

    if amount > balance:
        raise Exception("Insufficient balance!")

    balance -= amount

except ValueError as e:
    print("Invalid input:", e)

except Exception as e:
    print(e)

else:
    print("Withdrawal successful!")
    print(f"Remaining balance: {balance:.2f}")

finally:
    print("ATM transaction completed.")
```

```python
balance = 3000

try:
    price = float(input("Enter product price: "))

    if price <= 0:
        raise ValueError("Price must be positive.")

    if price > balance:
        raise InsufficientBalanceError("Insufficient balance!")

except ValueError as e:
    print("Invalid input:", e)

except InsufficientBalanceError as e:
    print(e)

else:
    balance -= price
    print("Payment successful!")
    print(f"Remaining balance: {balance:.2f}")

finally:
    print("Thank You!")
```

```python
marks = []

try:
    for i in range(5):
        mark = int(input(f"Enter mark for subject {i + 1}: "))

        if mark < 0 or mark > 100:
            raise ValueError("Marks must be between 0 and 100.")

        marks.append(mark)

except ValueError as e:
    print("Invalid mark:", e)

else:
    total = sum(marks)
    average = total / 5

    print("Marks:", marks)
    print("Total:", total)
    print("Average:", average)

finally:
    print("Marks entry process completed.")
```

```python
available_seats = 5

try:
    seats = int(input("Enter number of seats to book: "))

    if seats <= 0:
        raise ValueError("Enter a positive number of seats.")

    if seats > available_seats:
        raise Exception("Requested seats are unavailable.")

except ValueError as e:
    print("Invalid input:", e)

except Exception as e:
    print(e)

else:
    available_seats -= seats
    print("Booking successful!")
    print("Seats remaining:", available_seats)

finally:
    print("Booking process completed.")
```

```python
basic_salary = float(input("Enter basic salary : "))
working_days = int(input("Enter working days : "))

try:
    if basic_salary <= 0:
        raise ValueError("Salary must be postive!")

    if working_days <= 0:
        raise Exception("Working days must be positive!")

    daily_salary = basic_salary / working_days

except ValueError as e:
    print("Invalid Input:", e)

except Exception as e:
    print(e)

else:
    print("Daily salary:", daily_salary)

finally:
    print("Salary calculation completed.")
```
