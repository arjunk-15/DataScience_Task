```python
import numpy as np
import pandas as pd

data = {
    "Student_Name" : ["Arjun krish","Jenish","Nishanth"],
    "Student_Age" : [21, 22, 26],
    "Student_Mark" : [93, 0, None]
}
df=pd.DataFrame(data)
df.rename(
    columns={
        "Student_Name" : "Name",
        "Student_Age" : "Age",
        "Student_Mark" : "Mark"
    }, inplace=True
)
print(df)
```

```python
df=pd.DataFrame(data)
df["Student_Mark"] = df["Student_Mark"].replace(0, 35)
print(df)
```

```python
df["Student_Mark"]=df["Student_Mark"].astype(float)
print(df.dtypes)
```

```python
print(df.sort_values("Student_Mark", ascending=False))
```

```python
df["Result"]=np.where(df["Student_Mark"]>=40, "Pass", "Fail")
print(df)
```

```python
df["Student_Mark"]=df["Student_Mark"].fillna(0)
print(df)
```

```python
df["Student_Name"]=df["Student_Name"].str.upper()
print(df)
```

```python
df["Department"]=["BME", "BME", "IT"]
result = df.groupby("Department")["Student_Mark"].mean()
print(result)
```
