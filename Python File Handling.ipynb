```python
with open("students.txt", "w") as file:
    file.write("Arun,85\n")
    file.write("Priya,90\n")
    file.write("Kumar,78\n")

with open("students.txt", "r") as file:
    print("Student Records:")
    print(file.read())
```

```python
emp_id = input("Enter employee ID: ")
name = input("Enter employee name: ")
salary = input("Enter salary: ")

with open("employees.txt", "a") as file:
    file.write(f"{emp_id},{name},{salary}\n")

print("Employee details saved!")

try:
    with open("employees.txt", "r") as file:
        print("\nEmployee Records:")
        print(file.read())
except FileNotFoundError:
    print("No employee records found.")
```

```python
with open("employees.txt", "r") as file:
  content = file.read()

print("Word :",len(content.split()))
print("Lines :",len(content.splitlines()))
print("Characters :",len(content))
```

```python
with open("notes.txt", "a") as file:
  file.write("Hello, ")
  new_content = input("Enter new content: ")
  file.write(new_content + "\n" )

with open("notes.txt", "r") as file:
  print(file.read())
```

```python
with open("userx.txt", "a") as file:
  file.write("arjunkrish@05,12345")

name = input("Enter username :")
word = input("Enter password :")

with open("userx.txt", "r") as file:
  for line in file:
    user_name,password = line.strip().split(",")

    if user_name == name and password == word:
      print()
      print("Login successfully!")
    else:
      print("Invalid login details!")
```
