```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X, Y, color="red", linewidth=3)
plt.xticks(rotation=45)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Line Plot")
plt.show()
```

```python
student = ["Arjun", "Jenish", "Nishanth", "Subika", "Ruba"]
marks = [98, 65, 73, 83, 92]

plt.bar(student, marks)
plt.show()
```

```python
plt.barh(student, marks)
plt.show()
```

```python
plt.pie(
    marks,
    labels=student,
    autopct="%1.2f%%")
plt.show()
```

```python
X = [1, 2, 3, 4, 5]
Y = [5, 10, 15, 20, 25]

plt.scatter(X, Y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Scatter Plot")
plt.show()
```

```python
age = [43, 53, 32, 53, 21, 12, 96, 32, 21, 95]

plt.hist(age, bins=10)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("the Age Distribution")

plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(X, Y)
plt.title("Line chart")

plt.subplot(1,2,2)
plt.bar(X, Y)
plt.title("Bar chart")

plt.show()
```
