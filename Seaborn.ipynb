```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

Study_Hours = [12, 3, 8, 10]
Exam_Marks = [98, 32, 76, 85]

sns.scatterplot(x=Study_Hours, y=Exam_Marks)

plt.xlabel("Study Hours")
plt.ylabel("Exam Marks")

plt.show()
```

```python
data = {
    "Departments" : ["IT", "IT", "IT", "IT", "HR", "HR", "HR", "HR", "Finance", "Finance", "Finance", "Finance"],
    "Salary" : [30000, 40000, 45000, 90000, 25000, 35000, 40000, 50000, 40000, 50000, 60000, 150000]
}

df = pd.DataFrame(data)
sns.boxplot(data=df, x="Departments", y="Salary")

plt.title("Salary distribution")
plt.show()
```

```python
marks = {
    "Maths" : [98, 34, 76, 75, 100],
    "Science" : [85, 54, 97, 75, 52],
    "English" : [92, 75, 86, 83, 72],
    "Computer" : [65, 86, 64, 97, 98]
}

df = pd.DataFrame(marks)

sns.heatmap(df.corr(),
            annot= True,
            cmap="coolwarm",
            linewidths=0.5)

plt.title("Correlation Matrix")
plt.show()
```
